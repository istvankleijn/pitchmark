```python
import pathlib

import geopandas as gpd
import pandas as pd
import shapely

from pitchmark.physics.rolling import *
```

```python
folder = pathlib.Path().cwd() / "USGS_LIDAR"
```

```python
mesh_from_csv = pd.read_csv(folder / "redbud_mesh.csv", index_col=0)
mesh_geom = gpd.GeoSeries.from_wkt(mesh_from_csv["geometry"])
mesh_gdf = gpd.GeoDataFrame(data=mesh_from_csv, geometry=mesh_geom)
```

```python
df_from_csv = pd.read_csv(folder / "redbud_gdf.csv", index_col=0)
hole_geom = gpd.GeoSeries.from_wkt(df_from_csv["geometry"])
hole_gdf = gpd.GeoDataFrame(data=df_from_csv, geometry=hole_geom)
```

```python
mesh_gdf
```

```python
green_mesh_gdf = mesh_gdf.overlay(
    hole_gdf[hole_gdf.course_area == "putting_green"], how="intersection"
)
```

```python
x0, y0 = -690.0, -337.0
```

```python
redbud_green = Green(12.0, hole_location=(x0, y0), gdf=green_mesh_gdf,)
```
